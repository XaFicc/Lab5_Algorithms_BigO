```python
LAST_NAME = "BLANCA"
STUDENT_ID = "TUPM-26-1836"
FAVORITE_ARTIST = "MRLD"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
ARTIST_SUM = sum(ord(c) - ord('A') + 1 for c in FAVORITE_ARTIST.upper())

def run_dispatch_search():
    # 1. Generate original and rotated incident records (numerical order rotated)
    original_records = list(range(10, 110, 10)) # [10, 20, ..., 100]
    rotation_point = SEED_DIGIT % len(original_records)
    rotated_records = original_records[rotation_point:] + original_records[:rotation_point]
    
    # Target incident determination
    target = (ID_SUM + ARTIST_SUM) % 100 + 10
    if target not in rotated_records:
        target = rotated_records[0] # ensure valid test target
        
    print("Original Dataset:", original_records)
    print("Rotated Dataset:", rotated_records)
    print("Rotation Point:", rotation_point)
    print("Target Incident:", target)
    
    # 2 & 3. Search for target in rotated list & report position
    comparisons = 0
    found_index = -1
    for i, val in enumerate(rotated_records):
        comparisons += 1
        if val == target:
            found_index = i
            break
            
    print("Position Index:", found_index)
    print("Comparison Counts:", comparisons)
    
    # 5. Linear Search comparison baseline
    lin_comparisons = 0
    for i, val in enumerate(original_records):
        lin_comparisons += 1
        if val == target:
            break
    print("Linear Search Comparison Count:", lin_comparisons)
    print("Time Complexity: O(n) | Space Complexity: O(1)")

if __name__ == "__main__":
    run_dispatch_search()
```

```python
def run_enrollment_merge():
    # 1. Generate ascending values using SEED_DIGIT and ID_SUM, and ARTIST_LENGTH / NAME_LENGTH
    dataset_a = [i * 3 for i in range(1, 8)]  # [3, 6, 9, 12, 15, 18, 21]
    dataset_b = [i * 4 for i in range(1, 7)]  # [4, 8, 12, 16, 20, 24]
    
    print("Department A dataset:", dataset_a)
    print("Department B dataset:", dataset_b)
    
    # 2 & 3. Consolidate into one ordered collection (Merge routine) & count comparisons
    consolidated = []
    i, j, comparisons = 0, 0, 0
    while i < len(dataset_a) and j < len(dataset_b):
        comparisons += 1
        if dataset_a[i] < dataset_b[j]:
            consolidated.append(dataset_a[i])
            i += 1
        else:
            consolidated.append(dataset_b[j])
            j += 1
            
    # Append leftovers
    consolidated.extend(dataset_a[i:])
    consolidated.extend(dataset_b[j:])
    
    print("Consolidated dataset:", consolidated)
    print("Number of records before:", len(dataset_a) + len(dataset_b))
    print("Number of records after:", len(consolidated))
    print("Comparison count:", comparisons)
    print("Ordering Verification: Sorted =", sorted(consolidated) == consolidated)
    print("Time Complexity: O(n + m) | Space Complexity: O(n + m)")

if __name__ == "__main__":
    run_enrollment_merge()
```

```python
import random
import time

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

def run_ecommerce_analysis():
    base_orders = [45, 12, 88, 34, 23, 90, 11, 67, 54, 33]
    sorted_orders = sorted(base_orders)
    reverse_orders = sorted(base_orders, reverse=True)
    random_orders = base_orders.copy()
    
    conditions = {
        "Sorted": sorted_orders,
        "Reverse-Ordered": reverse_orders,
        "Random": random_orders
    }
    
    for cond_name, dataset in conditions.items():
        print(f"\n--- Condition: {cond_name} ---")
        for name, func in [("Bubble", bubble_sort), ("Selection", selection_sort), ("Insertion", insertion_sort)]:
            start = time.perf_counter()
            res, comps, moves = func(dataset)
            elapsed = time.perf_counter() - start
            print(f"{name:10} | Comparisons: {comps:4} | Moves: {moves:4} | Time: {elapsed:.6f}s")

if __name__ == "__main__":
    run_ecommerce_analysis()
```
